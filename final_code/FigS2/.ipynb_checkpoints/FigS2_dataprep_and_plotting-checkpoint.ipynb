```python
import warnings

with warnings.catch_warnings(record=True):
    df_li = []
    keys = []
    for file in filenames:
        file_path = os.path.join(exp_path, file)
        # display(file)
        df_file = pd.read_csv(file_path, sep='\t',encoding= 'unicode_escape', skiprows =[i for i in range(0,200)])
        # display(df_file)
        new_df= df_file.iloc[:,0:5]
        #display(new_df.head())
        #display(new_df.columns[2])
        strain_name = new_df.columns[2]
        #display(strain_name)
        if "MMY" in strain_name:
            new_df['strain'] = "WT"
        elif "YCY002_1" in strain_name:
            new_df['strain'] = "Δccr4"
        elif "YCY006_1" in strain_name:
            new_df['strain'] = "Δwhi5 Δccr4"

        if (new_df.iloc[4,1] == "1.34816"):
            new_df['sizerange'] = "10to250"
        elif (new_df.iloc[4,1]) == "2.40218":
            new_df['sizerange'] = "328to1000"
        new_df['filename'] = strain_name
        new_df.columns = ['Bin Number', 'Bin Diameter (Lower) - um', 'Diff. Number', 'Diff. Number/ mL', 'Diff. Volume %', 'strain', 'sizerange', 'filename']

        new_df.drop([0, 1,2], inplace = True)
        new_df.reset_index(inplace = True, drop = True)
        #display(new_df)

        list = [0,5,6,7,8]
        new_df_2= df_file.iloc[:,list]
        strain_name = new_df_2.columns[2]
        if "MMY" in strain_name:
            new_df_2['strain'] = "WT"
        elif "YCY002_1" in strain_name:
            new_df_2['strain'] = "Δccr4"
        elif "YCY006_1" in strain_name:
            new_df_2['strain'] = "Δwhi5 Δccr4"

        if (new_df_2.iloc[4,1] == "1.34816"):
            new_df_2['sizerange'] = "10to250"
        elif (new_df_2.iloc[4,1]) == "2.40218":
            new_df_2['sizerange'] = "328to1000"
        new_df_2['filename'] = strain_name
        #display(new_df_2.iloc[4,1])
        new_df_2.columns = ['Bin Number', 'Bin Diameter (Lower) - um', 'Diff. Number', 'Diff. Number/ mL', 'Diff. Volume %', 'strain', 'sizerange', 'filename']

        new_df_2.drop([0, 1,2], inplace = True)
        new_df_2.reset_index(inplace = True, drop = True)
        #display(new_df_2)

        list = [0,9,10,11,12]
        new_df_3= df_file.iloc[:,list]
        strain_name = new_df_3.columns[2]
        if "MMY" in strain_name:
            new_df_3['strain'] = "WT"
        elif "YCY002_1" in strain_name:
            new_df_3['strain'] = "Δccr4"
        elif "YCY006_1" in strain_name:
            new_df_3['strain'] = "Δwhi5 Δccr4"

        if (new_df_3.iloc[4,1] == "1.34816"):
            new_df_3['sizerange'] = "10to250"
        elif (new_df_3.iloc[4,1]) == "2.40218":
            new_df_3['sizerange'] = "328to1000"
        new_df_3['filename'] = strain_name
        #display(new_df_3.iloc[4,1])
        new_df_3.columns = ['Bin Number', 'Bin Diameter (Lower) - um', 'Diff. Number', 'Diff. Number/ mL', 'Diff. Volume %', 'strain', 'sizerange', 'filename']

        new_df_3.drop([0, 1,2], inplace = True)
        new_df_3.reset_index(inplace = True, drop = True)
        #display(new_df_3)

        list = [0,13,14,15,16]
        new_df_4= df_file.iloc[:,list]
        strain_name = new_df_4.columns[2]
        if "MMY" in strain_name:
            new_df_4['strain'] = "WT"
        elif "YCY002_1" in strain_name:
            new_df_4['strain'] = "Δccr4"
        elif "YCY006_1" in strain_name:
            new_df_4['strain'] = "Δwhi5 Δccr4"

        if (new_df_4.iloc[4,1] == "1.34816"):
            new_df_4['sizerange'] = "10to250"
        elif (new_df_4.iloc[4,1]) == "2.40218":
            new_df_4['sizerange'] = "328to1000"
        new_df_4['filename'] = strain_name
        #display(new_df_4.iloc[4,1])
        new_df_4.columns = ['Bin Number', 'Bin Diameter (Lower) - um', 'Diff. Number', 'Diff. Number/ mL', 'Diff. Volume %', 'strain', 'sizerange', 'filename']

        new_df_4.drop([0, 1,2], inplace = True)
        new_df_4.reset_index(inplace = True, drop = True)
        #display(new_df_4)

        timepoint_df = pd.concat([new_df, new_df_2, new_df_3, new_df_4])

        #  create amazing df
        df_li.append(timepoint_df)

big_df = pd.concat(df_li).reset_index(drop = True)
#print(big_df.filename.unique())
# display(big_df[big_df['strain'] == "whi5_ccr4_doubledel"])
```

# ***calculate bin cell vol from bin diameter*** 

```python
display(big_df.head())
big_df['Bin Diameter (Lower) - um'] = (big_df['Bin Diameter (Lower) - um']).astype(float)
big_df['Diff. Number'] = (big_df['Diff. Number']).astype(float)
big_df['Bin radius um'] = (big_df['Bin Diameter (Lower) - um'])/2
big_df['Bin cell_vol_fl'] = (4/3)*(math.pi)*((big_df['Bin radius um'])**3)
# display(big_df.head())
```

# ***assign rep as a separate column*** 

```python
# assign rep as a separate column
import warnings

with warnings.catch_warnings(record=True):
    big_df['rep'] = "rep"
    for row in big_df.index:
        if "rep1" in big_df['filename'].loc[row]:
            big_df['rep'].loc[row] = "rep1"
        else:
            big_df['rep'].loc[row] = "rep2"
```

# ***calclate binsize (vol) for each row (two sizeranges, so two unique values expected)*** 

```python

import warnings

with warnings.catch_warnings(record=True):
    list_2 =[]
    for filename in big_df['filename'].unique():
        file_df = big_df[big_df['filename'] == filename]
        # if filename == "MMY_rep1_SCD_100ul_31546_10to250fl.=#Z2":
        #     display(file_df)
        file_df['binsize_vol_fl'] = np.nan
        for row in file_df.index:
            if row != file_df.index[-1]:
                file_df['binsize_vol_fl'].loc[row] = (file_df['Bin cell_vol_fl'].loc[row + 1]) - (file_df['Bin cell_vol_fl'].loc[row])
        #print(file_df)
        # if filename == "MMY_rep1_SCD_100ul_31546_10to250fl.=#Z2":
        #     display(file_df)
        #     display(file_df['binsize_vol_fl'])
        file_df['mean_binsize_vol'] = np.nanmean(file_df['binsize_vol_fl'])
        #print(file_df)
        list_2.append(file_df)
    final_df = pd.concat(list_2).reset_index(drop = True)
    display(final_df.mean_binsize_vol.unique())
```

# ***combine both size ranges for each rep of each strain*** 

```python
# for the bigger size range, divide bin freq by the conversion factor calculated below
import warnings

with warnings.catch_warnings(record=True):
    list4 = []
    for filename in final_df['filename'].unique():
            file_df = final_df[final_df['filename'] == filename]
            for sizerange in file_df['sizerange'].unique():
                size_df = file_df[file_df['sizerange'] == sizerange]
                if sizerange == "328to1000":
                    size_df.drop((size_df[size_df['Bin Diameter (Lower) - um'] < 8.54]).index, inplace=True)
                elif sizerange == "10to250":
                    size_df.drop((size_df[size_df['Bin Diameter (Lower) - um'] > 8.54]).index, inplace=True)
                list4.append(size_df)


    final_df = pd.concat(list4).reset_index(drop = True)
    # conversion factor = mean_binsize_328to1000 / mean_binsize_10to250 
    # obtain these values^ from previous cell
    conversion_factor = 7.2578399/1.28282606
    
    final_df['corrected_bin_freq'] = np.nan
    for row in final_df.index:
        if final_df['sizerange'].loc[row] == "328to1000":
            final_df['corrected_bin_freq'].loc[row] =(final_df['Diff. Number'].loc[row])/conversion_factor
        elif final_df['sizerange'].loc[row] == "10to250":
            final_df['corrected_bin_freq'].loc[row] =(final_df['Diff. Number'].loc[row])
```

# ***calculate true mean cell volume, stdev and CV for each strain for reps individually and also for reps combined*** 

```python
# calculate mean cell volume, CV of cell volume and size distribution for each strain (Reps combined)
import warnings

with warnings.catch_warnings(record=True):
    
    list_3 = []
    for strain in final_df['strain'].unique():
        strain_df = final_df[final_df['strain'] == strain]
        for rep in strain_df['rep'].unique():
            rep_df = strain_df[strain_df['rep'] == rep]
            sum_diff_no = np.sum(rep_df['corrected_bin_freq'])
            rep_df['rep_no_percent'] = ((rep_df['corrected_bin_freq'])/sum_diff_no)*100
            rep_df['rep_bin_midpoint_cell_vol'] = ((rep_df['Bin cell_vol_fl'])+((rep_df['mean_binsize_vol'])/2))
            rep_df['rep_bin_midpoint_x_corrected_bin_freq'] = (rep_df['rep_bin_midpoint_cell_vol'])*(rep_df['corrected_bin_freq'])
            rep_df['rep_true_mean_cell_vol_fl'] = (np.sum(rep_df['rep_bin_midpoint_x_corrected_bin_freq']))/sum_diff_no
            rep_df['rep_midpt_minus_mean_wholesquared'] = ((rep_df['rep_bin_midpoint_cell_vol']) - (rep_df['rep_true_mean_cell_vol_fl']))**2
            rep_df['rep_midpt_minus_mean_wholesquared_x_freq'] = ((rep_df['rep_midpt_minus_mean_wholesquared'])*(rep_df['corrected_bin_freq']))
            rep_df['rep_summation_midpt_minus_mean_wholesquared_x_freq'] = np.sum(rep_df['rep_midpt_minus_mean_wholesquared_x_freq'])
            rep_df['rep_stdev'] = np.sqrt((rep_df['rep_summation_midpt_minus_mean_wholesquared_x_freq'])/(sum_diff_no-1))
            rep_df['rep_CV'] = (rep_df['rep_stdev'])/(rep_df['rep_true_mean_cell_vol_fl'])
            list_3.append(rep_df)
    final_df = pd.concat(list_3).reset_index(drop=True)
        
    list_5 = []
    for strain in final_df['strain'].unique():
        strain_df = final_df[final_df['strain'] == strain]
        sum_diff_no = np.sum(strain_df['corrected_bin_freq'])
        strain_df['sum_differential_num'] = sum_diff_no
        strain_df['bin_midpoint_cell_vol'] = ((strain_df['Bin cell_vol_fl'])+((strain_df['mean_binsize_vol'])/2))
        strain_df['bin_midpoint_x_corrected_bin_freq'] = (strain_df['bin_midpoint_cell_vol'])*(strain_df['corrected_bin_freq'])
        strain_df['true_mean_cell_vol_fl'] = (np.sum(strain_df['bin_midpoint_x_corrected_bin_freq']))/sum_diff_no
        strain_df['midpt_minus_mean_wholesquared'] = ((strain_df['bin_midpoint_cell_vol']) - (strain_df['true_mean_cell_vol_fl']))**2
        strain_df['midpt_minus_mean_wholesquared_x_freq'] = ((strain_df['midpt_minus_mean_wholesquared'])*(strain_df['corrected_bin_freq']))
        strain_df['summation_midpt_minus_mean_wholesquared_x_freq'] = np.sum(strain_df['midpt_minus_mean_wholesquared_x_freq'])
        strain_df['stdev'] = np.sqrt((strain_df['summation_midpt_minus_mean_wholesquared_x_freq'])/(sum_diff_no-1))
        strain_df['CV'] = (strain_df['stdev'])/(strain_df['true_mean_cell_vol_fl'])
        strain_df['sterr'] = (strain_df['stdev'])/(np.sqrt(strain_df['sum_differential_num']))

        list_5.append(strain_df)
    final_df = pd.concat(list_5).reset_index(drop=True)
    #display(final_df.head(30))
```
