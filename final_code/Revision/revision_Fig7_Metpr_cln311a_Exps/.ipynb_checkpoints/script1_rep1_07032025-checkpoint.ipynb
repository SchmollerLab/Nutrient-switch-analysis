# ***assign switch properties, complete phase, complete cycle and phase volume added*** 

```python
# assign post-switch generation_num to each generation in overall_df_with_rel
# if cycle started and ended before frame 50, post-switch gen = np.nan
# if cycle started after frame 50 and cell had seen switch, post-switch gen = generation_num - gen_at_switch
# if cycle started after frame 50 and cell had not seen switch, post-switch gen = generation_num 
# if cycle started before switch and ended after or at switch, post-switch gen = "switch_gen"
# also assign if each phase in each cycle saw switch or not (phase_saw_switch)
# save as overall_df_with_rel_and_switch_info
pd.set_option('display.max_rows', 350)
pd.set_option('display.max_columns', 100)
pd.set_option('display.width', 300)

overall_df_with_rel['post_switch_gen'] = np.nan
overall_df_with_rel['cycle_saw_switch'] = False
overall_df_with_rel['phase_saw_switch'] = False
overall_df_with_rel['cell_saw_switch'] = False
overall_df_with_rel['cell_saw_switch_as_bud'] = False
overall_df_with_rel['cell_saw_switch_in_G1'] = False
overall_df_with_rel['cell_saw_switch_in_S'] = False
overall_df_with_rel['complete_phase'] = 0
overall_df_with_rel['complete_cycle'] = 0
overall_df_with_rel['phase_length'] = 0
overall_df_with_rel['phase_volume_added'] = 0
list_for_joining = []

import warnings

with warnings.catch_warnings(record=True):

    for strain in overall_df_with_rel['strain'].unique():
        strain_df = overall_df_with_rel[overall_df_with_rel['strain'] == strain]
        for pos in (strain_df['position']).unique():
            pos_df = strain_df[strain_df["position"] == pos]
            pos_max_frame = pos_df['frame_i'].max()
            pos_min_frame = pos_df['frame_i'].min()
            for cell in (pos_df['Cell_ID']).unique():
                cell_df = pos_df[pos_df["Cell_ID"] == cell]
                if 50 in (cell_df['frame_i'].values):
                    switch_df = cell_df[cell_df['frame_i'] == 50]
                    gen_at_switch = switch_df['generation_num'].iloc[0]
                    stage_at_switch = switch_df['cell_cycle_stage'].iloc[0]
                    cell_df['cell_saw_switch'] = True
                    if gen_at_switch == 0:
                        cell_df['cell_saw_switch_as_bud'] = True
                    if (stage_at_switch == "G1"):
                        cell_df['cell_saw_switch_in_G1'] = True
                    else:
                        cell_df['cell_saw_switch_in_S'] = True
                else:
                    gen_at_switch = np.nan
                for cycle in (cell_df['generation_num']).unique(): 
                    cycle_df = cell_df[cell_df["generation_num"] == cycle]
                    cycle_df = cycle_df.sort_values(by = 'frame_i')
                    if ((cycle_df['frame_i'].min()) > 50) & (np.isnan(gen_at_switch) == False):
                        cycle_df['post_switch_gen'] = cycle - gen_at_switch
                    elif ((cycle_df['frame_i'].min()) > 50) & (np.isnan(gen_at_switch) == True):
                        cycle_df['post_switch_gen'] = cycle
                    if 50 in (cycle_df['frame_i'].values):
                        cycle_df['cycle_saw_switch'] = True
                        cycle_df['post_switch_gen'] = "switch_gen"
                    if ((cycle_df['frame_i'].min()) > pos_min_frame) & ((cycle_df['frame_i'].max()) < pos_max_frame):
                        cycle_df['complete_cycle'] = 1
                    if sum(cycle_df['is_cell_dead']) > 0:
                        cycle_df['complete_cycle'] = 0


                    for phase in (cycle_df['cell_cycle_stage'].unique()):
                        phase_df = cycle_df[cycle_df['cell_cycle_stage'] == phase]
                        phase_df['phase_length'] = (len(phase_df))-1
                        phase_df['phase_volume_added'] = ((phase_df['cell_vol_fl'].iloc[-1]) - (phase_df['cell_vol_fl'].iloc[0]))
                        if 50 in (phase_df['frame_i'].values):
                            phase_df['phase_saw_switch'] = True
                        #if cell dies in cycle, change complete_phase to 0
                        if ((phase_df['frame_i'].min()) > pos_min_frame) & ((phase_df['frame_i'].max()) < pos_max_frame):
                            phase_df['complete_phase'] = 1
                            phase_df['does_cell_die_in_this_phase'] = 0
                        if sum(phase_df['is_cell_dead']) > 0:
                            phase_df['complete_phase'] = 0
                            phase_df['does_cell_die_in_this_phase'] = 1
                        else:
                            phase_df['does_cell_die_in_this_phase'] = 0
                        list_for_joining.append(phase_df)
overall_df_with_rel_and_switch_info = pd.concat(list_for_joining)                    

 


```


# ***Filtering for daughters that were buds at switch and form the first wave of daughters of S/G2/M switchers post-switch*** 

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 400)

# calculate a unique cell-cycle id accross files by appending strain, pos, cell id (and gen)
overall_df_with_rel_and_switch_info['f_pos_cell_id'] = overall_df_with_rel_and_switch_info.apply(lambda x: f'{x["strain"]}_{x["position"]}_Cell_{x["Cell_ID"]}_Gen_{int(x["generation_num"])}', axis=1)

print(overall_df_with_rel_and_switch_info.shape)


# select only cycles that faced switch
switch_cycles_df = overall_df_with_rel_and_switch_info[(overall_df_with_rel_and_switch_info['cycle_saw_switch'] == True) 
                                                       & (overall_df_with_rel_and_switch_info['complete_phase'] == 1)
                                                      ]

##################################################################
#####       WAVE 1 OF DAUGHTERS POST-SWITCH       ################
##################################################################
# from all switch-facing cycles, extract only S-phases of mothers that faced switch
# from these switch-facing S-phases, extract relative IDs to get first wave of daughters that were born after switch
post_switch_daughters_wave1_series = []
for cell_cycle in ((switch_cycles_df['f_pos_cell_id']).unique()):
    cell_df = switch_cycles_df[switch_cycles_df['f_pos_cell_id'] == cell_cycle]
    S_switcher_df = cell_df[(cell_df['cell_cycle_stage'] == "S") & (cell_df['phase_saw_switch'] == True) & (cell_df['cell_saw_switch_as_bud'] == False)]
    if (S_switcher_df.empty == False):
        first_daughter = (S_switcher_df['relative_ID'].min())
        pos = S_switcher_df['position'].iloc[0]
        strain = S_switcher_df['strain'].iloc[0]
        cell_id = S_switcher_df['Cell_ID'].iloc[0]
        post_switch_daughters_wave1_cycle_df = pd.DataFrame({'strain': [strain], 'position': [pos],  'cell': [cell_cycle], 'relative_ID': [first_daughter], 'Cell_ID': [cell_id]})
        post_switch_daughters_wave1_series.append(post_switch_daughters_wave1_cycle_df)
post_switch_daughters_wave1_df = pd.concat(post_switch_daughters_wave1_series).reset_index(drop = True)



# find the cells that are post switch first wave daughters in the overall df and assign post_switch_daughter_wave = 1 to all their rows.
# later on for filtering for first post-switch gen of these first wave daughters, use (post_switch_daughter_wave = 1) alongside (post_switch_gen = 1)
overall_df_with_rel_and_switch_info['S_switch_cells_daughter_wave'] = np.nan

for idx in post_switch_daughters_wave1_df.index:
    pos = post_switch_daughters_wave1_df['position'].loc[idx]
    strain = post_switch_daughters_wave1_df['strain'].loc[idx]
    rel = post_switch_daughters_wave1_df['relative_ID'].loc[idx]
    overall_df_with_rel_and_switch_info.loc[((overall_df_with_rel_and_switch_info['position'] == pos) 
                                             & (overall_df_with_rel_and_switch_info['strain'] == strain)
                                             & (overall_df_with_rel_and_switch_info['Cell_ID'] == rel) 
                                            ), 'S_switch_cells_daughter_wave'] = 1

# # get number of first wave daughters, filtered for using (switch_cells_daughter_wave = 1) alongside (post_switch_gen = 1) 
# first_gen_first_wave_daughters_post_switch_df = overall_df_with_rel_and_switch_info[(overall_df_with_rel_and_switch_info['S_switch_cells_daughter_wave'] == 1) 
#                                                                                     & ((overall_df_with_rel_and_switch_info['post_switch_gen'] == 1)|
#                                                                                        (overall_df_with_rel_and_switch_info['post_switch_gen'] == "1"))
#                                                                                     & (overall_df_with_rel_and_switch_info['complete_phase'] == 1)
#                                                                                      ]
# #display(first_gen_first_wave_daughters_post_switch_df)
# print("switch daughters excl dead", len(first_gen_first_wave_daughters_post_switch_df['f_pos_cell_id'].unique()))

```

```python
##################################################################
#####       WAVE 2 OF DAUGHTERS POST-SWITCH       ################
##################################################################
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 400)

# from all mtoher S-switchers, get first S-phase after switch (post_switch_gen == 1)
# from these S-phases, extract relative IDs to get second wave of daughters that were born after switch

post_switch_daughters_wave2_series = []
for cell_cycle in ((overall_df_with_rel_and_switch_info['f_pos_cell_id']).unique()):
    cell_df = overall_df_with_rel_and_switch_info[overall_df_with_rel_and_switch_info['f_pos_cell_id'] == cell_cycle]
    switch_cells_first_S_post_switch_df = cell_df[((cell_df['post_switch_gen'] == "1")|(cell_df['post_switch_gen'] == 1)) 
                                                & (cell_df['cell_saw_switch'] == True)
                                                & (cell_df['cell_cycle_stage'] == "S") 
                                                & (cell_df['complete_phase'] == 1)  
                                                & (cell_df['cell_saw_switch_as_bud'] == False)
                                                & (cell_df['cell_saw_switch_in_S'] == 1)
                                                 ]
    if (switch_cells_first_S_post_switch_df.empty == False):
        second_daughter = (switch_cells_first_S_post_switch_df['relative_ID'].min())
        pos = switch_cells_first_S_post_switch_df['position'].iloc[0]
        strain = switch_cells_first_S_post_switch_df['strain'].iloc[0]
        post_switch_daughters_wave2_cycle_df = pd.DataFrame({'strain': [strain], 'position': [pos],  'cell': [cell_cycle], 'relative_ID': [second_daughter]})
        post_switch_daughters_wave2_series.append(post_switch_daughters_wave2_cycle_df)
post_switch_daughters_wave2_df = pd.concat(post_switch_daughters_wave2_series).reset_index(drop = True)


# find the cells that are post switch second wave daughters in the overall df and assign post_switch_daughter_wave = 2 to all their rows.
# later on for filtering for first post-switch gen of these second wave daughters, use (post_switch_daughter_wave = 2) alongside (post_switch_gen = 1)
for idx in post_switch_daughters_wave2_df.index:
    pos = post_switch_daughters_wave2_df['position'].loc[idx]
    strain = post_switch_daughters_wave2_df['strain'].loc[idx]
    rel = post_switch_daughters_wave2_df['relative_ID'].loc[idx]
    overall_df_with_rel_and_switch_info.loc[((overall_df_with_rel_and_switch_info['position'] == pos) 
                                             & (overall_df_with_rel_and_switch_info['strain'] == strain)
                                             & (overall_df_with_rel_and_switch_info['Cell_ID'] == rel) 
                                            ), 'S_switch_cells_daughter_wave'] = 2 
        
# # get number of second wave daughters, filtered for using (switch_cells_daughter_wave = 2) alongside (post_switch_gen = 1)  
# first_gen_second_wave_daughters_post_switch_df = overall_df_with_rel_and_switch_info[(overall_df_with_rel_and_switch_info['S_switch_cells_daughter_wave'] == 2) 
#                                                                                     & ((overall_df_with_rel_and_switch_info['post_switch_gen'] == 1)|
#                                                                                        (overall_df_with_rel_and_switch_info['post_switch_gen'] == "1"))
#                                                                                     & (overall_df_with_rel_and_switch_info['complete_phase'] == 1)]
                

# print("daughters 2 excl dead cells", len((first_gen_second_wave_daughters_post_switch_df['f_pos_cell_id']).unique()))

# #display(first_gen_second_wave_daughters_post_switch_df['f_pos_cell_id'].unique())
```


# ***Filtering for fourth wave of daughters of S/G2/M switchers***

```python
##################################################################
#####       WAVE 4 OF DAUGHTERS POST-SWITCH       ################
##################################################################

# from all mtoher S-switchers, get third S-phase after switch (post_switch_gen == 3)
# from these S-phases, extract relative IDs to get fourth wave of daughters that were born after switch
post_switch_daughters_wave4_series = []
for cell_cycle in ((overall_df_with_rel_and_switch_info['f_pos_cell_id']).unique()):
    cell_df = overall_df_with_rel_and_switch_info[overall_df_with_rel_and_switch_info['f_pos_cell_id'] == cell_cycle]
    switch_cells_third_S_post_switch_df = cell_df[((cell_df['post_switch_gen'] == "3")|(cell_df['post_switch_gen'] == 3)) 
                                                  & (cell_df['cell_saw_switch'] == True) 
                                                  & (cell_df['cell_cycle_stage'] == "S")
                                                  & (cell_df['complete_phase'] == 1)  
                                                  & (cell_df['cell_saw_switch_as_bud'] == False)
                                                  & (cell_df['cell_saw_switch_in_S'] == 1)]
    
    if (switch_cells_third_S_post_switch_df.empty == False):
        fourth_daughter = (switch_cells_third_S_post_switch_df['relative_ID'].min())
        pos = switch_cells_third_S_post_switch_df['position'].iloc[0]
        strain = switch_cells_third_S_post_switch_df['strain'].iloc[0]
        post_switch_daughters_wave4_cycle_df = pd.DataFrame({'strain': [strain], 'position': [pos],  'cell': [cell_cycle], 'relative_ID': [fourth_daughter]})
        post_switch_daughters_wave4_series.append(post_switch_daughters_wave4_cycle_df)
post_switch_daughters_wave4_df = pd.concat(post_switch_daughters_wave4_series).reset_index(drop = True)


# find the cells that are post switch fourth wave daughters in the overall df and assign post_switch_daughter_wave = 4 to all their rows.
# later on for filtering for first post-switch gen of these fourth wave daughters, use (post_switch_daughter_wave = 4) alongside (post_switch_gen = 1)
for idx in post_switch_daughters_wave4_df.index:
    pos = post_switch_daughters_wave4_df['position'].loc[idx]
    strain = post_switch_daughters_wave4_df['strain'].loc[idx]
    rel = post_switch_daughters_wave4_df['relative_ID'].loc[idx]
    overall_df_with_rel_and_switch_info.loc[((overall_df_with_rel_and_switch_info['position'] == pos) 
                                             & (overall_df_with_rel_and_switch_info['strain'] == strain)
                                             & (overall_df_with_rel_and_switch_info['Cell_ID'] == rel) 
                                            ), 'S_switch_cells_daughter_wave'] = 4 
    
# # get number of third wave daughters, filtered for using (switch_cells_daughter_wave = 4) alongside (post_switch_gen = 1)  
# first_gen_fourth_wave_daughters_post_switch_df = overall_df_with_rel_and_switch_info[(overall_df_with_rel_and_switch_info['S_switch_cells_daughter_wave'] ==4) 
#                                                                                     & ((overall_df_with_rel_and_switch_info['post_switch_gen'] == 1)|
#                                                                                        (overall_df_with_rel_and_switch_info['post_switch_gen'] == "1"))
#                                                                                     & (overall_df_with_rel_and_switch_info['complete_phase'] == 1)
#                                                                                    ]
# print("daughters 4 excl dead cells", len(first_gen_fourth_wave_daughters_post_switch_df['f_pos_cell_id'].unique()))
```

# ***Assigning frames_since_cycle_start for each row for each cycle***
# ***+ Assigning daughterhood_binary for each row for each cycle***
# ***+ Assigning birth_or_div_frame, bud_emerg_frame, div_frame for each cycle***
# ***+ Assigning birth_or_div_vol, bud_emerg_sys_vol, div_sys_vol for each cycle***

```python

import warnings

with warnings.catch_warnings(record=True):

    unique_f_pos_cell_id = []
    for fposid in ((overall_df_with_rel_and_switch_info['f_pos_cell_id']).unique()):
        fposid_df = overall_df_with_rel_and_switch_info[overall_df_with_rel_and_switch_info['f_pos_cell_id'] == fposid]
        fposid_df = fposid_df.sort_values(by=['frame_i'])
        if (len(fposid_df['Cell_ID'].unique())) > 1:
            print("HELP", fposid_df)
        ############## daughterhood_binary assignment ################
        cycle_gen_num =fposid_df['generation_num'].iloc[0]
        fposid_df['daughterhood_binary'] = np.nan
        if cycle_gen_num == 1:
            fposid_df['daughterhood_binary'] = 1
        else:
            fposid_df['daughterhood_binary'] = 0
            
        ############## frames_since_cycle_start assignment ################   
        cycle_start_frame =fposid_df['frame_i'].iloc[0]
        fposid_df['frames_since_cycle_start'] = np.nan
        for row in fposid_df.index:
            fposid_df['frames_since_cycle_start'].loc[row] = (fposid_df['frame_i'].loc[row]) - cycle_start_frame
            
        ############## birth_or_div_frame + vol, BE, div ################ 
        if (fposid_df['complete_phase'].iloc[0] == 1):
            fposid_df['birth_or_div_frame'] = fposid_df['frame_i'].iloc[0]
            fposid_df['birth_or_div_vol'] = fposid_df['cell_vol_fl'].iloc[0] 
            s_df = fposid_df[fposid_df['cell_cycle_stage'] == "S"]
            s_df = s_df.sort_values(by=['frame_i'])
            if (len(s_df) != 0):
                if (s_df['complete_phase'].iloc[0] == 1):
                    fposid_df['bud_emerg_frame'] = s_df['frame_i'].iloc[0]
                    fposid_df['bud_emerg_sys_vol'] = s_df['combined_mother_bud_volume'].iloc[0]
                    fposid_df['div_frame'] = s_df['frame_i'].iloc[-1]
                    fposid_df['div_sys_vol'] = s_df['combined_mother_bud_volume'].iloc[-1]
                else:
                    fposid_df['bud_emerg_frame'] = np.nan
                    fposid_df['bud_emerg_sys_vol'] = np.nan
                    fposid_df['div_frame'] = np.nan
                    fposid_df['div_sys_vol'] = np.nan
            else:
                fposid_df['bud_emerg_frame'] = np.nan
                fposid_df['bud_emerg_sys_vol'] = np.nan
                fposid_df['div_frame'] = np.nan
                fposid_df['div_sys_vol'] = np.nan
        else:
            fposid_df['birth_or_div_frame'] = np.nan
            fposid_df['birth_or_div_vol'] = np.nan
            s_df = fposid_df[fposid_df['cell_cycle_stage'] == "S"]
            s_df = s_df.sort_values(by=['frame_i'])
            if (len(s_df) != 0):
                if (s_df['complete_phase'].iloc[0] == 1):
                    fposid_df['bud_emerg_frame'] = s_df['frame_i'].iloc[0]
                    fposid_df['bud_emerg_sys_vol'] = s_df['combined_mother_bud_volume'].iloc[0]
                    fposid_df['div_frame'] = s_df['frame_i'].iloc[-1]
                    fposid_df['div_sys_vol'] = s_df['combined_mother_bud_volume'].iloc[-1]
                else:
                    fposid_df['bud_emerg_frame'] = np.nan
                    fposid_df['bud_emerg_sys_vol'] = np.nan
                    fposid_df['div_frame'] = np.nan
                    fposid_df['div_sys_vol'] = np.nan
            else:
                fposid_df['bud_emerg_frame'] = np.nan
                fposid_df['bud_emerg_sys_vol'] = np.nan
                fposid_df['div_frame'] = np.nan
                fposid_df['div_sys_vol'] = np.nan

        
        
        unique_f_pos_cell_id.append(fposid_df)
overall_df_with_rel_and_switch_info = pd.concat(unique_f_pos_cell_id)
```
