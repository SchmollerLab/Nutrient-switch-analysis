```python
channels = ['mCitrine_Yagya', 'mScarlet_Ph3_Yagya', 'Dia_Ph3_mCherry']
# for every channel, calculate amount from mother and bud cells combined
for ch in channels:
    try:
        overall_df[f'{ch}_combined_amount_mother_bud_cell'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_amount_dataPrepBkgr_cell']
                + x.loc[f'{ch}_amount_dataPrepBkgr_cell_rel']
                if (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc[f'{ch}_amount_dataPrepBkgr_cell_rel']) == False)
                else x.loc[f'{ch}_amount_dataPrepBkgr_cell']
                ),
            axis=1
        )
        overall_df[f'{ch}_combined_amount_mother_bud_nuc'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_amount_dataPrepBkgr_nuc']
                + x.loc[f'{ch}_amount_dataPrepBkgr_nuc_rel']
                if x.loc['cell_cycle_stage_sa']=='S' and (np.isnan(x.loc[f'{ch}_amount_dataPrepBkgr_nuc_rel']) == False)
                else x.loc[f'{ch}_amount_dataPrepBkgr_nuc']
                ),
            axis=1
        )
        
    except KeyError:
        continue
        
```

```python
#combine mother and bud nuclear and cellular volumes.
#for cytoplasm, subtract mother_bud_combined_nuclear from mother_bud_combined_cellular

overall_df['combined_mother_bud_volume_cell'] = overall_df.apply(
    lambda x: x.loc['cell_vol_fl_cell']+x.loc['cell_vol_fl_cell_rel'] if\
    (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc['cell_vol_fl_cell_rel']) == False) else\
    x.loc['cell_vol_fl_cell'],
    axis=1
)
overall_df['combined_mother_bud_volume_nuc'] = overall_df.apply(
    lambda x: x.loc['cell_vol_fl_nuc']+x.loc['cell_vol_fl_nuc_rel'] if\
    (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc['cell_vol_fl_nuc_rel']) == False) else\
    x.loc['cell_vol_fl_nuc'],
    axis=1
)
overall_df['combined_mother_bud_volume_cytoplasm'] = overall_df.apply(
    lambda x: x.loc['combined_mother_bud_volume_cell']-x.loc['combined_mother_bud_volume_nuc'],
    axis=1
)

#for each compartment (excluding cytoplasm),
#calculate combined_mother_bud_area_pixels by adding area in pixels
    
#for cellular compartment, calculate combined_mother_bud_area_pixels_cell
overall_df['combined_mother_bud_area_pixels_cell'] = overall_df.apply(
    lambda x: (
        x.loc['cell_area_pxl_cell']
        + x.loc['cell_area_pxl_cell_rel']
        if (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc['cell_area_pxl_cell_rel']) == False) 
        else x.loc['cell_area_pxl_cell']
        ),
    axis=1
)
#for cellular compartment, calculate combined_mother_bud_area_pixels_nuc
overall_df['combined_mother_bud_area_pixels_nuc'] = overall_df.apply(
    lambda x: (
        x.loc['cell_area_pxl_nuc']
        + x.loc['cell_area_pxl_nuc_rel']
        if (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc['cell_area_pxl_nuc_rel']) == False) 
        else x.loc['cell_area_pxl_nuc']
        ),
    axis=1
)
```

```python
 
#for each channel
    #for each compartment (excluding cytoplasm),
        #calculate sum pixel intensity for mother and bud by multiplying mean pixel intensity by area in pixels
        #calculate combined_mother_bud_sum_pixel_intensity by adding sum pixel intensities for mother and bud
        #calculate combined_mother_bud_mean_pixel_intensity by dividing combined_mother_bud_sum_pixel_intensity by combined_mother_bud_area_pixels


for ch in channels:
    try:
        #for cellular compartment, calculate sum_pixel_intensity of mother
        overall_df[f'{ch}_sum_pixel_intensity_cell'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_mean_cell']
                * x.loc['cell_area_pxl_cell']
                ),
            axis=1
        )
        #for cellular compartment, calculate sum_pixel_intensity of bud
        overall_df[f'{ch}_sum_pixel_intensity_cell_rel'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_mean_cell_rel']
                * x.loc['cell_area_pxl_cell_rel']
                ),
            axis=1
        )
        #for nuclear compartment, calculate sum_pixel_intensity of mother
        overall_df[f'{ch}_sum_pixel_intensity_nuc'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_mean_nuc']
                * x.loc['cell_area_pxl_nuc']
                ),
            axis=1
        )
        #for nuclear compartment, calculate sum_pixel_intensity of bud
        overall_df[f'{ch}_sum_pixel_intensity_nuc_rel'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_mean_nuc_rel']
                * x.loc['cell_area_pxl_nuc_rel']
                ),
            axis=1
        )
        #for cellular compartment, calculate combined_mother_bud_sum_pixel_intensity_cell
        overall_df[f'{ch}_combined_mother_bud_sum_pixel_intensity_cell'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_sum_pixel_intensity_cell']
                + x.loc[f'{ch}_sum_pixel_intensity_cell_rel']
                if (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc[f'{ch}_sum_pixel_intensity_cell_rel']) == False) 
                else x.loc[f'{ch}_sum_pixel_intensity_cell']
                ),
            axis=1
        )
        #for nuclear compartment, calculate combined_mother_bud_sum_pixel_intensity_nuc
        overall_df[f'{ch}_combined_mother_bud_sum_pixel_intensity_nuc'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_sum_pixel_intensity_nuc']
                + x.loc[f'{ch}_sum_pixel_intensity_nuc_rel']
                if (x.loc['cell_cycle_stage_sa']=='S') and (np.isnan(x.loc[f'{ch}_sum_pixel_intensity_nuc_rel']) == False) 
                else x.loc[f'{ch}_sum_pixel_intensity_nuc']
                ),
            axis=1
        )
        #for cellular compartment, calculate combined_mother_bud_mean_pixel_intensity_cell
        overall_df[f'{ch}_combined_mother_bud_mean_pixel_intensity_cell'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_combined_mother_bud_sum_pixel_intensity_cell']
                / x.loc['combined_mother_bud_area_pixels_cell']
                if x.loc['cell_cycle_stage_sa']=='S'
                else x.loc[f'{ch}_mean_cell']
                ),
            axis=1
        )
        #for nuclear compartment, calculate combined_mother_bud_mean_pixel_intensity_nuc
        overall_df[f'{ch}_combined_mother_bud_mean_pixel_intensity_nuc'] = overall_df.apply(
            lambda x: (
                x.loc[f'{ch}_combined_mother_bud_sum_pixel_intensity_nuc']
                / x.loc['combined_mother_bud_area_pixels_nuc']
                if x.loc['cell_cycle_stage_sa']=='S'
                else x.loc[f'{ch}_mean_nuc']
                ),
            axis=1
        )
        
    except KeyError:
        continue

```

```python
 
#for each channel
    #for each compartment (excluding cytoplasm),
        #calculate combined_mother_bud_median_pixel_intensity by averaging median_pixel_intensity of mother and bud

for ch in channels:
    try:
        #for cellular compartment, calculate combined_mother_bud_median_pixel_intensity_cell
        overall_df[f'{ch}_combined_mother_bud_median_pixel_intensity_cell'] = overall_df.apply(
            lambda x: (
                np.nanmean([x.loc[f'{ch}_median_cell'], x.loc[f'{ch}_median_cell_rel']])
                if x.loc['cell_cycle_stage_sa']=='S'
                else x.loc[f'{ch}_median_cell']
                ),
            axis=1
        )
        #for nuclear compartment, calculate combined_mother_bud_median_pixel_intensity_nuc
        overall_df[f'{ch}_combined_mother_bud_median_pixel_intensity_nuc'] = overall_df.apply(
            lambda x: (
                np.nanmean([x.loc[f'{ch}_median_nuc'], x.loc[f'{ch}_median_nuc_rel']])
                if x.loc['cell_cycle_stage_sa']=='S'
                else x.loc[f'{ch}_median_nuc']
                ),
            axis=1
        )

    except KeyError:
        continue

```

# ***for each complete pre-switch cell cycle, calculate time relative to bud emergence*** 

```python
complete_preswitch_cycles = overall_df[(overall_df['growthmedium_sa'] == "SCD")
                            & (overall_df['cycle_saw_switch_sa'] == False)
                            & (overall_df['complete_cycle_sa'] == True)
                            & (overall_df['relationship_sa'] != "bud")]

complete = []
import warnings

with warnings.catch_warnings(record=True):
    for rep in complete_preswitch_cycles['rep'].unique():
        rep_df = complete_preswitch_cycles[complete_preswitch_cycles['rep'] == rep]
        for cellcycle in rep_df['f_pos_cell_id_sa'].unique():
            cycle_df = rep_df[rep_df['f_pos_cell_id_sa'] == cellcycle]
            cycle_df.sort_values(by=['frame_i'], inplace = True)

            S_df = cycle_df[cycle_df['cell_cycle_stage_sa'] == "S"]
            S_df.sort_values(by=['frame_i'], inplace = True)
            bud_emerg_frame = S_df.frame_i.min() 
            cycle_df['bud_emerg_frame'] = bud_emerg_frame 

            for frame in cycle_df['frame_i'].unique():
                frame_ind = (cycle_df[cycle_df['frame_i'] == frame]).index 
                cycle_df.loc[frame_ind[0], 'frame_rel_to_bud_emergence'] = frame - bud_emerg_frame 

            complete.append(cycle_df)



new_complete_preswitch_cycles_df = pd.concat(complete).reset_index(drop = True)
new_complete_preswitch_cycles_df['time_rel_to_bud_emergence_mins'] = new_complete_preswitch_cycles_df['frame_rel_to_bud_emergence']*3

display(new_complete_preswitch_cycles_df.head(15))
            
        
        
```

```python
import statsmodels.api as sm
import warnings
with_p = []

with warnings.catch_warnings(record=True):
     for rep in overall_df['rep'].unique():
        rep_df = overall_df[(overall_df['rep'] == rep) & (overall_df['strain_sa'] != "MMY")]
        for cellcycle in rep_df['f_pos_cell_id_sa'].unique():
            cycle_df = rep_df[rep_df['f_pos_cell_id_sa'] == cellcycle]
            if (cycle_df['complete_cycle_sa'].iloc[0] == 1) & (cycle_df['relationship_sa'].iloc[0] != "bud"):
                G1_df = cycle_df[cycle_df['cell_cycle_stage_sa'] == "G1"]
                if (len(G1_df) != 0):
                    
                    fit1_params = sm.OLS.from_formula(formula='sys_mScarlet_Ph3_Yagya_amount_dataPrepBkgr_cell~time_in_cell_cycle_mins_sa', 
                                    data = G1_df ).fit().params
                    p1 = fit1_params[0]
                    last_10_frames_df = cycle_df.nlargest(10, ['time_in_cell_cycle_sa']) 
                    fit2_params = sm.OLS.from_formula(formula='sys_mScarlet_Ph3_Yagya_amount_dataPrepBkgr_cell~time_in_cell_cycle_mins_sa', 
                                    data = last_10_frames_df ).fit().params
                    p2 = fit2_params[0]
                    cycle_df['mScarlet_p1'] = p1
                    cycle_df['mScarlet_p2'] = p2
                else:
                    display(rep, cellcycle)
                    cycle_df['mScarlet_p1'] = np.nan
                    cycle_df['mScarlet_p2'] = np.nan
            else:
                cycle_df['mScarlet_p1'] = np.nan
                cycle_df['mScarlet_p2'] = np.nan

            with_p.append(cycle_df)
overall_df_with_Ps = pd.concat(with_p).reset_index(drop = True)
display(overall_df_with_Ps.head(10))
```
