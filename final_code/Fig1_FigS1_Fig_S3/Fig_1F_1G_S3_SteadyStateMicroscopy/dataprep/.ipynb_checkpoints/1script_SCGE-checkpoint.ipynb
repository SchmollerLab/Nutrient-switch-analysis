
# ***Define binit*** 

```python
def binit(x, y, maxval, numbins):
    binning_data = []
    bins = np.linspace(0, maxval, numbins)
    bins2 = bins+0.5*(bins[2]-bins[1])
    #print (bins, bins2)
    #print(x)
    bin = np.digitize(x, bins)
    #print (bin)
    #mu = np.full_like(bins, np.nan)
    #error = np.full_like(bins, np.nan)
    #print (mu, error)

    for k in range(1, numbins):
        ind = np.where(bin==k)[0].tolist()





        # print (y.index)
        # print (y.iloc[61])

        if ((sum(bin==k))>=3):
            # print (k, (y.iloc[ind]))
            #print (y[(ind[0]):(ind[-1]+1)])
            mu = np.nanmean(y.iloc[ind])
            error = np.nanstd(y.iloc[ind])/np.sqrt(sum(~np.isnan(y.iloc[ind])))
            stdev = np.nanstd(y.iloc[ind])
            n = sum(~np.isnan(y.iloc[ind]))
            # print (mu, error, stdev, n)

            #print (mu, error)
            #mu = np.nanmean(y[(ind[0]):(ind[-1]+1)])
            # error = np.nanstd(y[(ind[0]):(ind[-1]+1)])/np.sqrt(sum(~np.isnan(y[(ind[0]):(ind[-1]+1)])))
            binning_d = [k, mu, error, numbins, bins.round(2), bins2.round(2), stdev, n]
            binning_data.append(binning_d)

    co= ['k', 'mu', 'error', 'numbins', 'bins', 'bins2', 'stdev', 'n']
    binning_data = pd.DataFrame(binning_data, columns = co)

    return binning_data


sns.set_theme(style="darkgrid")
```
