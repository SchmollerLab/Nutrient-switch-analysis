# ***assign custom cell cycle phase based on mother-bud-volume ratio*** 

```python
granddata['phase_yc'] = "phase_yc"

import warnings

with warnings.catch_warnings(record=True):
    for ind in granddata.index:
        if ((granddata['ratio_volumes_bud_moth']).iloc[ind]) == 0:
            granddata['phase_yc'].iloc[ind] = "G1"
        else:
            if ((((granddata['ratio_volumes_bud_moth']).iloc[ind]) > 0) & (((granddata['ratio_volumes_bud_moth']).iloc[ind]) <= 0.3)):
                granddata['phase_yc'].iloc[ind] = "S"
            else:
                granddata['phase_yc'].iloc[ind] = "G2M"


#display(granddata)
```
